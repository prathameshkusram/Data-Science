```python
import numpy as np
import pandas as pd

#created the sample dataset with missing value
data={
    'A':[1,2,100,4,5],
    'B':[3,np.nan,7,8,9],
    'C':[np.nan,12,13,14,15],
    'D':[16,17,18,np.nan,20]

}
df=pd.DataFrame(data)
print("Original DataFrame")
print(df)
```

```python
#to check how many null values are there in data frame
df.isnull().sum()
```

```python
# Droping rows with null values
df.dropna()
```

```python
#droping columns with null values
df.dropna(axis=1)
```

```python
df['B']
```

```python
df['B'].fillna(df['B'].mean())
```

```python
df['B'].fillna(df['B'].median())
```

```python
# if you want to fill any constant
df['C'].fillna(0)
```

```python
df['C'].fillna('SomeStrings')
```

```python
sns.histplot(df.age,kde=True)
```

```python
#whenever you have normal distribution data and no outlier inpute with mean
df['mean_imputation']=df['age'].fillna(df['age'].mean())
```

```python
df[['age','mean_imputation']]
```

```python
sns.distplot(df['mean_imputation'],kde=1)
```

```python
# Mean imputation works well with normally distributed data /no outlier is present
```
