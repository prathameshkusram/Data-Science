```python
import pandas as pd
import numpy as np

# Main dataset (NFHS)
df_nfhs = pd.read_csv("/content/datafile.csv")

# Hospital dataset
df_hosp = pd.read_csv("/content/HospitalsInIndia.csv")
```

```python
def clean_value(val):
    if pd.isna(val):
        return np.nan
    s = str(val).strip()
    if s == '*':
        return np.nan
    if s.startswith('(') and s.endswith(')'):
        s = s[1:-1]
    try:
        return float(s)
    except:
        return np.nan

for col in df_nfhs.columns[2:]:
    df_nfhs[col] = df_nfhs[col].apply(clean_value)

df_nfhs.fillna(df_nfhs.median(numeric_only=True), inplace=True)
```

```python
# Standardize column names
df_hosp.columns = df_hosp.columns.str.lower()

# Clean state names
df_hosp['state'] = df_hosp['state'].str.strip().str.lower()
df_nfhs['State/UT'] = df_nfhs['State/UT'].str.strip().str.lower()
```

```python
# Count hospitals per state
hosp_count = df_hosp.groupby('state').size().reset_index(name='hospital_count')
```

```python
df = pd.merge(df_nfhs, hosp_count,
              left_on='State/UT',
              right_on='state',
              how='left')
```

```python
df['hospital_count'].fillna(df['hospital_count'].median(), inplace=True)
```

```python
print(df.columns.tolist())
```

```python
df.rename(columns={
    "Population living in households with electricity (%)": "electricity",
    "Population living in households with an improved drinking-water source1 (%)": "water",
    "Population living in households that use an improved sanitation facility2 (%)": "sanitation",
    "Households using clean fuel for cooking3 (%)": "clean_fuel",
    "Women (age 15-49) who are literate4 (%)": "literacy",
    "Households with any usual member covered under a health insurance/financing scheme (%)": "insurance",

    "Children age 12-23 months fully vaccinated based on information from either vaccination card or mother's recall11 (%)": "vaccination",

    "Children under 5 years who are stunted (height-for-age)18 (%)": "stunting",

    "Non-pregnant women age 15-49 years who are anaemic (<12.0 g/dl)22 (%)": "anaemia"

}, inplace=True)
```

```python
features = [
    "electricity",
    "water",
    "sanitation",
    "clean_fuel",
    "literacy",
    "insurance",
    "vaccination",
    "stunting",
    "anaemia",
    "hospital_count"
]

df_model = df[features].copy()
```

```python
df.columns = df.columns.str.strip()
```

```python
score = (
    (df_model['electricity'] < df_model['electricity'].median()).astype(int) +
    (df_model['sanitation'] < df_model['sanitation'].median()).astype(int) +
    (df_model['water'] < df_model['water'].median()).astype(int) +
    (df_model['insurance'] < df_model['insurance'].median()).astype(int) +
    (df_model['literacy'] < df_model['literacy'].median()).astype(int) +
    (df_model['vaccination'] < df_model['vaccination'].median()).astype(int) +
    (df_model['stunting'] > df_model['stunting'].median()).astype(int) +
    (df_model['anaemia'] > df_model['anaemia'].median()).astype(int) +
    (df_model['hospital_count'] < df_model['hospital_count'].median()).astype(int)
)

df_model['target'] = (score >= 4).astype(int)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X = df_model.drop('target', axis=1)
y = df_model['target']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print("Train size:", X_train.shape)
print("Test size:", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

models = {
    "Logistic Regression": LogisticRegression(max_iter=500),
    "Decision Tree": DecisionTreeClassifier(max_depth=6),
    "Random Forest": RandomForestClassifier(n_estimators=100),
    "SVM": SVC()
}

results = []

for name, model in models.items():
    model.fit(X_train, y_train)
    pred = model.predict(X_test)

    acc = accuracy_score(y_test, pred)
    prec = precision_score(y_test, pred)
    rec = recall_score(y_test, pred)
    f1 = f1_score(y_test, pred)

    results.append([name, acc, prec, rec, f1])

    print(f"{name} → Acc:{acc:.2f} Prec:{prec:.2f} Rec:{rec:.2f} F1:{f1:.2f}")
```

```python
import pandas as pd

results_df = pd.DataFrame(results, columns=[
    "Model","Accuracy","Precision","Recall","F1"
])

print("\nModel Comparison:\n", results_df)

best_model = results_df.sort_values("F1", ascending=False).iloc[0]
print("\nBest Model:\n", best_model)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

kmeans = KMeans(n_clusters=3, random_state=42)
labels_km = kmeans.fit_predict(X_scaled)

print("KMeans Silhouette:", silhouette_score(X_scaled, labels_km))
```

```python
from sklearn.cluster import AgglomerativeClustering

agg = AgglomerativeClustering(n_clusters=3)
labels_agg = agg.fit_predict(X_scaled)

print("Agglomerative Silhouette:", silhouette_score(X_scaled, labels_agg))
```

```python
from sklearn.cluster import DBSCAN

db = DBSCAN(eps=2.5, min_samples=5)
labels_db = db.fit_predict(X_scaled)

# Ignore noise (-1)
valid = labels_db != -1

if len(set(labels_db[valid])) > 1:
    print("DBSCAN Silhouette:", silhouette_score(X_scaled[valid], labels_db[valid]))
else:
    print("DBSCAN did not form proper clusters")
```

```python
from sklearn.metrics import davies_bouldin_score

print("KMeans DB Index:", davies_bouldin_score(X_scaled, labels_km))
print("Agglomerative DB Index:", davies_bouldin_score(X_scaled, labels_agg))
```
