```python
# ==============================
# PART A: Data Understanding & Profiling
# ==============================

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ------------------------------
# Step 1: Load Dataset
# ------------------------------
df = pd.read_csv("healthcare_dataset.csv")

print("\nFirst 5 Records:")
print(df.head())

# ------------------------------
# Step 2: Dataset Shape
# ------------------------------
rows, cols = df.shape
print("\nNumber of Rows:", rows)
print("Number of Columns:", cols)

# ------------------------------
# Step 3: Feature Types
# ------------------------------
numeric_features = df.select_dtypes(include=["int64", "float64"]).columns
categorical_features = df.select_dtypes(include=["object"]).columns

print("\nNumeric Features:", list(numeric_features))
print("Categorical Features:", list(categorical_features))

# ------------------------------
# Step 4: Missing Values
# ------------------------------
print("\nMissing Values Per Column:")
print(df.isnull().sum())

# ------------------------------
# Step 5: Statistical Summary
# ------------------------------
print("\nStatistical Summary:")
print(df.describe())

# ------------------------------
# Step 6: SEPARATE Box Plot for Each Numeric Feature
# ------------------------------
for col in numeric_features:
    plt.figure()
    plt.boxplot(df[col].dropna())
    plt.title(f"Box Plot of {col}")
    plt.ylabel(col)
    plt.grid(True)

# ------------------------------
# Step 7: Bar Chart for Each Categorical Feature
# ------------------------------
for col in categorical_features:
    counts = df[col].value_counts()
    plt.figure()
    plt.bar(counts.index.astype(str), counts.values)
    plt.title(f"Bar Chart of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.grid(True)

# ------------------------------
# Step 8: Heatmap using Matplotlib (NO seaborn)
# ------------------------------
corr = df[numeric_features].corr()

plt.figure(figsize=(8,6))
plt.imshow(corr, cmap="coolwarm", interpolation="nearest")
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Correlation Heatmap of Numeric Features")

# ------------------------------
# FORCE DISPLAY OF ALL PLOTS
# ------------------------------
plt.show()
```
